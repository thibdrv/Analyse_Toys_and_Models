# Filters : Year / Month

**TOTAL : 2018-01 to 2020-02 / 26 month**

The two filters below are automatically applied to **all KPIs** in this notebook via df_filtre :

`YEAR` : Filters by a specific year (set to None to disable filtering).

`TRAILING_N_Months` : Keeps only the last N months of data (relative to the most recent date available after applying the YEAR filter). Set to None to disable.

Both filters can be combined (year 2019, then the last 3 months of that year).

```python
YEAR = None               # 2019, or None to not filter by year
TRAILING_N_MONTHS = 3     # 3 for the last 3 months, or None to keep all

df_filtered = df.copy()

if YEAR is not None:
    df_filtered = df_filtered[df_filtered["orderDate"].dt.year == YEAR]

if TRAILING_N_MONTHS is not None:
    max_date = df_filtered["orderDate"].max()
    start_date = max_date - pd.DateOffset(months=TRAILING_N_MONTHS)
    df_filtered = df_filtered[df_filtered["orderDate"] > start_date]

start_period, end_period = df_filtered["orderDate"].min(), df_filtered["orderDate"].max()

print(f"Filter applied = Year: {YEAR} | Rolling window : last {TRAILING_N_MONTHS} month(s) \n")
print(f"Resulting period : from {start_period.date()} to {end_period.date()} \n")
print(f"{len(df_filtered)} rows out of {len(df)} total ({len(df_filtered)/len(df)*100:.1f}%)")
```

# 1. Stock Below Critical Threshold : Identify products with low stock levels to prevent stockouts.

**Explanation :**

This KPI compares available inventory (quantityInStock) with average monthly consumption calculated over a rolling window (TRAILING_N_MONTHS). The Stock_coverage_months ratio indicates how many months current inventory will last based on observed consumption rates.

Over the analyzed period (Nov 2019 – Feb 2020), 1 product is at critical stock levels (1960 BSA Gold Star DBD34 : 0.35 months of coverage, or ~10 days) and 1 is on reorder alert (1968 Ford Mustang: 1.3 months).

**Decision Making :**

Only 2 out of 109 SKUs are under pressure, meaning the stockout risk is localized rather than systemic.

Recommended action : trigger a priority reorder for these 2 products (ensuring a supplier lead time of < 10 days for the critical stock item), and monitor this table monthly rather than taking action across the entire catalog.

```python
# Product-level aggregation
stock_kpi = (
    df_filtered.groupby("productName")
      .agg(
          category=("category", "first"),
          productScale=("productScale", "first"),
          quantityInStock=("quantityInStock", "first"),
          buyPrice=("buyPrice", "first"),
          MSRP=("MSRP", "first"),
          Total_quantity_ordered=("quantityOrdered", "sum"),
      )
      .reset_index()
)

# Monthly average order
stock_kpi["Monthly_avg_consumption"] = stock_kpi["Total_quantity_ordered"] / TRAILING_N_MONTHS

# Safety stock in months
stock_kpi["Stock_coverage_months"] = np.where(
    stock_kpi["Monthly_avg_consumption"] > 0,
    stock_kpi["quantityInStock"] / stock_kpi["Monthly_avg_consumption"],
    np.inf
)

# Stock_status_3months
conditions = [
    stock_kpi["Stock_coverage_months"] < 1,
    stock_kpi["Stock_coverage_months"] < 2,
]
choices = ["Critical stock", "Reorder warning"]
stock_kpi["Stock_status"] = np.select(conditions, choices, default="Sufficient stock")

# Stock valuation
stock_kpi["stock_value"] = (stock_kpi["quantityInStock"] * stock_kpi["buyPrice"]).round(2)

stock_kpi = stock_kpi.sort_values("Stock_coverage_months")
stock_kpi
```

```python
# Product Stock Distribution VS Status

status_counts = stock_kpi["Stock_status"].value_counts()

STATUS_COLORS = {"Sufficient stock": "#2ca02c", "Reorder warning": "#ff8c00", "Critical stock": "#d62728"}
colors = [STATUS_COLORS.get(s, "#999999") for s in status_counts.index]

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Donut chart
wedges, _, _ = axes[0].pie(
    status_counts, colors=colors, startangle=90,
    autopct=lambda p: f"{p:.1f}%" if p > 4 else "",
    pctdistance=0.80, wedgeprops=dict(width=0.45, edgecolor="white", linewidth=2),
    textprops=dict(color="white", fontsize=11, fontweight="bold"),
)
axes[0].set_title("Product distribution by stock status", fontsize=13, fontweight="bold", pad=15)
axes[0].legend(wedges, [f"{s}  ({n} · {n/status_counts.sum()*100:.1f}%)" for s, n in status_counts.items()],
               loc="upper center", bbox_to_anchor=(0.5, -0.02), frameon=False, fontsize=10)
axes[0].text(0, 0.08, f"{status_counts.sum()}", ha="center", fontsize=18, fontweight="bold")
axes[0].text(0, -0.12, "products", ha="center", fontsize=11, color="#555")

# Total stock by category
stock_by_category = stock_kpi.groupby("category")["quantityInStock"].sum().sort_values()
bars = axes[1].barh(stock_by_category.index, stock_by_category.values, color="#4c72b0")

axes[1].bar_label(bars, fmt="{:,.0f}", padding=3, fontsize=9)
axes[1].set_title("Total stock by category", fontsize=13, fontweight="bold", pad=15)
axes[1].set_xlabel("Quantity in stock")

plt.tight_layout()
plt.show()
```

# 2. Average Order Processing Time + Orders Above Average Delivery Time

**Explanation :**

The average preparation delay is 3.28 days across 48 usable orders. The calculated logistics service level (retard_livraison) is 100% : no delivered order exceeded its requiredDate.

The correlation between product scale (productScale) and preparation delay is virtually zero (Pearson = 0.10, non-significant for Spearman / Kendall), showing that product size does not explain variations in delay.

**Decision Making :**

Since the logistics service level is already at 100%, there are no urgent corrective actions needed for delivery delays. However, 44% of orders exceed the average delay of 3.28 days : investigate internal fulfillment processes (warehouse workload, day of the week) rather than product type, which is not a key driver. Separate point of attention : 6 orders are "On Hold" due to exceeded creditLimit this is a credit management issue rather than a logistics issue, and warrants a dedicated alert to the finance department.

_____



The CSV does not directly contain prep_delay or delivery_delay: they are recalculated from **orderDate**, **shippedDate**, and **requiredDate**.

Orders without a shippedDate (not yet shipped: **On Hold**, **Cancelled**) are excluded from the delay calculation, behaving like an implicit filter.

```python
df_filtered["shippedDate"].isnull().sum()
```

```python
logistics = df_filtered.dropna(subset=["shippedDate"]).copy()

# preparation_delay = shippedDate - orderDate (in days)
logistics["delai_preparation"] = (logistics["shippedDate"] - logistics["orderDate"]).dt.days

# delivery_delay = 1 if delivered after the required date, 0 otherwise
logistics["retard_livraison"] = (logistics["shippedDate"] > logistics["requiredDate"]).astype(int)

print(f"Usable orders (with shipping date): {len(logistics)} / {len(df_filtered)}")

logistics[["orderDate", "requiredDate", "shippedDate", "delai_preparation", "retard_livraison"]].head()
```

```python
delays_by_order = logistics.groupby("orderNumber")["delai_preparation"].first().reset_index()
delays_by_order
```

# 3. Stock Sell-Through Rate : Measure how quickly inventory is sold for each product.

**Explanation :**

105 out of 109 products (96%) are categorized as "Very slow turnover" (< 10% of inventory sold over the period), representing a total of 547,399 units in stock. Only a single product the exact same one facing a critical stockout shows a rate > 100%, driven by extremely low stock relative to demand.

**Decision Making :**

This very low turnover rate could indicate genuine overstocking or simply be an artifact of analyzing a short observation window (3 months) against inventory built up over a longer period a point to verify before taking action. If confirmed over a broader time frame, it calls for a review of purchasing policies (reducing order quantities) rather than launching an immediate clearance operation across 96% of the catalog.

**Sell Through Rate**

```python
# Extract product (category, buy price, MSRP)
sell_through = (
    df_filtered.groupby("productName")
    .agg(
        category=("category", "first"),
        buyPrice=("buyPrice", "first"),
        MSRP=("MSRP", "first"),
    )
    .reset_index()
)

sell_through["avg_stock_qty"] = sell_through["productName"].map(avgStockProducts)
sell_through["avg_order_qty"] = sell_through["productName"].map(avgStockOrdered)

sell_through["sell_through_rate"] = (
    (sell_through["avg_order_qty"] / sell_through["avg_stock_qty"].replace(0, np.nan))
    * 100
).round(2)

# Assign turnover status based on the sell-through rate
conditions = [
    sell_through["sell_through_rate"] < 10,
    sell_through["sell_through_rate"] < 30,
    sell_through["sell_through_rate"] < 60,
]
labels = ["Very slow turnover", "Slow turnover", "Moderate turnover"]

sell_through["turnover_status"] = np.select(
    conditions, labels, default="Fast turnover"
)

# Sort by ascending sell-through rate
sell_through = sell_through.sort_values("sell_through_rate").reset_index(drop=True)

sell_through
```

```python
print(sell_through["turnover_status"].value_counts())

fig, ax = plt.subplots(figsize=(8, 5))
order = ["Very slow turnover", "Slow turnover", "Moderate turnover", "Fast turnover"]
sns.countplot(
    data=sell_through,
    y="turnover_status",
    order=order,
    hue="turnover_status",
    palette=["#d62728", "#ff7f0e", "#00FF00", "#1E90FF"],
    legend=False,
    ax=ax,
)
ax.set_title("Product distribution by turnover speed")
ax.set_xlabel("Number of products")
ax.set_ylabel("Turnover status")
plt.tight_layout()
plt.show()
```

**Dead Stock Analysis: Locked-Up Capital vs. Liquidation Gain**

```python
# Very slow turnover products : potential liquidation gain, locked-up revenue, number of references
slow_turnover = sell_through[sell_through["turnover_status"] == "Very slow turnover"]

liquidation_gain = (slow_turnover["avg_stock_qty"] * slow_turnover["MSRP"] * 0.70).sum()
nb_slow_turnover = slow_turnover["productName"].nunique()
locked_revenue = (slow_turnover["avg_stock_qty"] * slow_turnover["buyPrice"]).sum()

print(f"Number of very slow turnover products       : {nb_slow_turnover} \n")
print(f"Revenue locked up in these products         : {locked_revenue:,.2f} € \n")
print(f"Potential liquidation gain (at 70% MSRP)    : {liquidation_gain:,.2f} € \n")
```

# 4. Late Delivery Rate : Identify logistics issues and improve delivery times.

Explanation :

Very slow-moving products tie up €30.07M in inventory purchase value. Clearing this stock at a 70% discount off MSRP would generate a theoretical liquidation gain of €38.79M.

______

```python
# Total number of late delivered orders
delayed_count = logistics["retard_livraison"].sum()

# Calculation of overall late delivery rate (%)
late_rate = (delayed_count / len(logistics)) * 100

print(f"Overall late delivery rate : {late_rate:.2f} % \n")
print(f"Number of delayed orders   : {delayed_count} out of {len(logistics)} delivered orders")
```

```python
# Distribution of late deliveries by product category and customer country
late_by_category = (
    logistics.groupby("category")["retard_livraison"]
    .apply(lambda x: (x > 0).sum())
    .sort_values(ascending=False)
)

# Use country_customer if available, otherwise fall back to country
country_col = "country_customer" if "country_customer" in logistics.columns else "country"

late_by_country = (
    logistics.groupby(country_col)["retard_livraison"]
    .apply(lambda x: (x > 0).sum())
    .sort_values(ascending=False)
    .head(10)
)

# Plot
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.barplot(x=late_by_category.values, y=late_by_category.index, ax=axes[0], color="#c44e52")
axes[0].set_title("Late orders by category")
axes[0].set_xlabel("Number of late orders")
axes[0].set_ylabel("Category")

sns.barplot(x=late_by_country.values, y=late_by_country.index, ax=axes[1], color="#c44e52")
axes[1].set_title("Top 10 countries by late orders")
axes[1].set_xlabel("Number of late orders")
axes[1].set_ylabel("Country")

plt.tight_layout()
plt.show()
```
