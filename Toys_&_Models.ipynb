# Finance

1. Highest / Lowest Revenue-Generating Customers : Identify the customers generating the most revenue to better retain them.

```python
Finance_KPI_1 = pd.read_sql("""
WITH
  pay_total AS (
    SELECT
        customerNumber,
        SUM(amount) AS total_Payer,
        paymentDate AS date_payment
    FROM payments
    GROUP BY paymentDate, customerNumber
  ),
  com_total AS (
    SELECT customerNumber
    FROM orders
    WHERE status != 'Cancelled'
    GROUP BY customerNumber
  )

SELECT
    c.customerName AS client,
    c.city,
    c.country,
    p.total_Payer,
    p.date_payment
FROM customers c
JOIN pay_total p ON c.customerNumber = p.customerNumber
JOIN com_total o ON c.customerNumber = o.customerNumber
ORDER BY c.customerName, p.date_payment DESC, p.total_Payer DESC;
""",
conn)

Finance_KPI_1
```

2. Receivables Collection Rate by Customer : Identify customers with a high amount of unpaid orders.

```python
Finance_KPI_2 = pd.read_sql("""
SELECT
    c.customerName,
    SUM(od.quantityOrdered * od.priceEach) AS Total_Commande,
    COALESCE(pay.totalPaid, 0) AS Total_Paye,
    ROUND(SUM(od.quantityOrdered * od.priceEach) - COALESCE(pay.totalPaid, 0), 2) AS Reste_A_Payer,
    ROUND((COALESCE(pay.totalPaid, 0) / SUM(od.quantityOrdered * od.priceEach)) * 100, 2) || ' %' AS Taux_Recouvrement,
    ROUND(c.creditLimit, 0) AS credit_limit,
    c.creditLimit - (SUM(od.quantityOrdered * od.priceEach) - COALESCE(pay.totalPaid, 0)) AS credit_restant,
    ROUND(
        CASE
            WHEN c.creditLimit > 0 THEN
                ((c.creditLimit - (SUM(od.quantityOrdered * od.priceEach) - COALESCE(pay.totalPaid, 0))) / c.creditLimit) * 100
            ELSE 0
        END, 2
    ) || ' %' AS credit_utilise

FROM customers c
JOIN orders o ON c.customerNumber = o.customerNumber
JOIN orderdetails od ON o.orderNumber = od.orderNumber
LEFT JOIN (
    SELECT customerNumber, SUM(amount) AS totalPaid
    FROM payments
    GROUP BY customerNumber
) pay ON c.customerNumber = pay.customerNumber

WHERE o.status IN ('Shipped', 'Resolved')

GROUP BY
    c.customerNumber,
    c.customerName,
    c.creditLimit,
    pay.totalPaid

HAVING (SUM(od.quantityOrdered * od.priceEach) - COALESCE(pay.totalPaid, 0)) > 0

ORDER BY Reste_A_Payer DESC;
""",
conn)

Finance_KPI_2
```

3. Quarterly Sales Growth : Identify seasonal trends or growth opportunities.

```python
Finance_KPI_3 = pd.read_sql("""
WITH QCategorie AS (
    SELECT  -- CA per Quarter
        p.productLine AS categorie,
        strftime('%Y', o.orderDate) AS annee,
        CAST((CAST(strftime('%m', o.orderDate) AS INTEGER) - 1) / 3 AS INTEGER) + 1 AS trimestre,
        SUM(od.quantityOrdered * od.priceEach) AS CA_actuel
    FROM orders o
    JOIN orderdetails od ON o.orderNumber = od.orderNumber
    JOIN products p ON od.productCode = p.productCode
    WHERE o.status IN ('Shipped', 'Resolved')
    GROUP BY p.productLine, annee, trimestre
)
SELECT
    annee, trimestre, categorie, ROUND(CA_actuel, 2) AS CA_actuel,
    -- Previous Quarter Comparison
    ROUND(LAG(CA_actuel) OVER (PARTITION BY categorie ORDER BY annee, trimestre), 2) AS CA_trim_precedent,
    ROUND(((CA_actuel - LAG(CA_actuel) OVER (PARTITION BY categorie ORDER BY annee, trimestre))
           / LAG(CA_actuel) OVER (PARTITION BY categorie ORDER BY annee, trimestre)) * 100, 2) || ' %' AS "Croissance_Q-1",
	  -- Year-over-Year
    ROUND(LAG(CA_actuel, 4) OVER (PARTITION BY categorie ORDER BY annee, trimestre), 2) AS "CA_annee-1",
    ROUND(((CA_actuel - LAG(CA_actuel, 4) OVER (PARTITION BY categorie ORDER BY annee, trimestre))
           / LAG(CA_actuel, 4) OVER (PARTITION BY categorie ORDER BY annee, trimestre)) * 100, 2) || ' %' AS "Croissance_A-1"
FROM QCategorie
ORDER BY categorie, annee DESC, trimestre DESC;
""",
conn)

Finance_KPI_3
```

4. Average Payment Amount + Customers Below Average : Assess customers’ payment capacity.

```python
Finance_KPI_4 = pd.read_sql("""
WITH CalculGlobal AS (
    SELECT
        c.customerName AS client,
        p.amount,
        p.paymentDate AS date,
        c.country AS pays,
        ROUND(AVG(p.amount) OVER(), 2) AS moy_generale,
        ROUND(AVG(p.amount) OVER(PARTITION BY c.customerNumber), 2) AS moy_client
    FROM customers c
    JOIN payments p ON c.customerNumber = p.customerNumber
)
SELECT
    date,
    client,
    pays,
    amount,
    moy_generale,
    ROUND(((amount - moy_generale) / moy_generale) * 100, 2) || ' %' AS "ecart_mg_%",
    ROUND(amount - moy_generale, 2) AS ecart_montant,
    moy_client,
    ROUND(moy_client - moy_generale, 2) AS ecart_client_moy,
    ROUND(((moy_client - moy_generale) / moy_generale) * 100, 2) || ' %' AS "ecart_client_%"
FROM CalculGlobal
ORDER BY date DESC;
""",
conn)

Finance_KPI_4
```

5. Payment Rate by Payment Delay : Identify how long customers take to pay after placing an order.

```python
Finance_KPI_5 = pd.read_sql("""
SELECT
    strftime('%Y', o.orderDate) AS annee,
    strftime('%m', o.orderDate) AS mois,
    o.orderNumber,
    c.customerName,
    o.orderDate,
    MIN(p.paymentDate) AS premier_paiement,
    julianday(MIN(p.paymentDate)) - julianday(o.orderDate) AS delai_pp,
    MAX(p.paymentDate) AS date_payer,
    julianday(MAX(p.paymentDate)) - julianday(o.orderDate) AS delai_total
FROM orders o
JOIN payments p ON o.customerNumber = p.customerNumber
JOIN customers c ON c.customerNumber = o.customerNumber
WHERE p.paymentDate >= o.orderDate -- On ignore les paiements passés AVANT la commande
GROUP BY o.orderNumber
ORDER BY annee DESC, mois DESC, delai_total DESC;
""",
conn)

Finance_KPI_5
```

# Sales

1. Monthly Revenue by Region + Monthly Growth Rate : Track revenue generated by region and month to identify geographic trends.

```python
Sales_KPI_1 = pd.read_sql("""
SELECT
    o.orderNumber,
    o.orderDate,
    o.status,
    c.customerNumber,
    c.country AS Country,
    COALESCE(c.state, c.country) AS Region, -- Utilise le state s'il existe, sinon le pays
    c.city AS City,
    od.productCode,
    od.quantityOrdered,
    od.priceEach,
    (od.quantityOrdered * od.priceEach) AS Revenue
FROM orders o
JOIN orderdetails od ON o.orderNumber = od.orderNumber
JOIN customers c ON o.customerNumber = c.customerNumber
WHERE o.status != 'Cancelled';
""",
conn)

Sales_KPI_1
```

2. Top / Bottom-Selling Products by Category : Identify the best-performing products in each category.

```python
Sales_KPI_2 = pd.read_sql("""
WITH product_sales AS (
    SELECT
        p.productLine,
        p.productCode,
        p.productName,
        SUM(od.quantityOrdered) AS total_quantity_sold,
        ROUND(SUM(od.quantityOrdered * od.priceEach), 2) AS total_revenue,
        DENSE_RANK() OVER (
            PARTITION BY p.productLine
            ORDER BY SUM(od.quantityOrdered * od.priceEach) DESC
        ) AS rank_in_category
    FROM products p
    JOIN orderdetails od ON p.productCode = od.productCode
    JOIN orders o ON od.orderNumber = o.orderNumber
    WHERE o.status != 'Cancelled' -- On exclut les commandes annulées
    GROUP BY p.productLine, p.productCode, p.productName
)
SELECT
    productLine,
    rank_in_category,
    productCode,
    productName,
    total_quantity_sold,
    total_revenue
FROM product_sales
ORDER BY productLine, rank_in_category;
""",
conn)

Sales_KPI_2
```

3. Gross Margin by Product and Category : Measure gross margin to identify the most and least profitable products and categories.

```python
Sales_KPI_3 = pd.read_sql("""
SELECT
    p.productLine,
    p.productCode,
    p.productName,
    SUM(od.quantityOrdered) AS total_quantity_sold,
    ROUND(SUM(od.quantityOrdered * od.priceEach), 2) AS total_revenue,
    ROUND(SUM(od.quantityOrdered * p.buyPrice), 2) AS total_cost,
    ROUND(SUM(od.quantityOrdered * (od.priceEach - p.buyPrice)), 2) AS gross_margin,
    ROUND(
        100.0 * SUM(od.quantityOrdered * (od.priceEach - p.buyPrice)) / SUM(od.quantityOrdered * od.priceEach),
        2
    ) AS margin_rate_pct
FROM products p
JOIN orderdetails od ON p.productCode = od.productCode
JOIN orders o ON od.orderNumber = o.orderNumber
WHERE o.status != 'Cancelled'
GROUP BY p.productLine, p.productCode, p.productName
ORDER BY gross_margin DESC; -- Trier par marge brute décroissante (ou margin_rate_pct DESC)
""",
conn)

Sales_KPI_3
```

4. Monthly Sales Growth Rate by Category : Measure the performance of each product category over time.

```python
Sales_KPI_4 = pd.read_sql("""
WITH monthly_category_sales AS (
    SELECT
        p.productLine,
        strftime('%Y-%m', o.orderDate) AS sales_month,
        ROUND(SUM(od.quantityOrdered * od.priceEach), 2) AS current_month_revenue
    FROM products p
    JOIN orderdetails od ON p.productCode = od.productCode
    JOIN orders o ON od.orderNumber = o.orderNumber
    WHERE o.status != 'Cancelled'
    GROUP BY p.productLine, strftime('%Y-%m', o.orderDate)
),
sales_with_lag AS (
    SELECT
        productLine,
        sales_month,
        current_month_revenue,
        LAG(current_month_revenue) OVER (
            PARTITION BY productLine
            ORDER BY sales_month
        ) AS previous_month_revenue
    FROM monthly_category_sales
)
SELECT
    productLine,
    sales_month,
    current_month_revenue,
    previous_month_revenue,
    ROUND(current_month_revenue - previous_month_revenue, 2) AS absolute_growth,
    ROUND(
        100.0 * (current_month_revenue - previous_month_revenue) / previous_month_revenue,
        2
    ) AS growth_rate_pct
FROM sales_with_lag
ORDER BY productLine, sales_month;
""",
conn)

Sales_KPI_4
```

5. Average Order Value (AOV) : Measure the average value of orders.

```python
pd.read_sql("""
WITH order_totals AS (
    SELECT
        o.orderNumber,
        c.country,
        SUM(od.quantityOrdered * od.priceEach) AS order_value
    FROM orders o
    JOIN orderdetails od ON o.orderNumber = od.orderNumber
    JOIN customers c ON o.customerNumber = c.customerNumber
    WHERE o.status != 'Cancelled'
    GROUP BY o.orderNumber, c.country
)
SELECT
    country,
    COUNT(orderNumber) AS total_orders,
    ROUND(AVG(order_value), 2) AS avg_order_value
FROM order_totals
GROUP BY country
ORDER BY avg_order_value DESC;
""",
conn)
```

```python
Sales_KPI_5 = pd.read_sql("""
SELECT
    o.orderNumber,
    o.orderDate,
    o.requiredDate,
    o.shippedDate,
    o.status,
    o.customerNumber,
    od.productCode,
    od.quantityOrdered,
    od.priceEach,
    p.buyPrice,
    p.productLine,
    (od.quantityOrdered * od.priceEach) AS Revenue,
    (od.quantityOrdered * p.buyPrice) AS Cost,
    (od.quantityOrdered * (od.priceEach - p.buyPrice)) AS GrossMargin,
    CASE
        WHEN o.shippedDate IS NOT NULL THEN (julianday(o.shippedDate) - julianday(o.orderDate))
        ELSE NULL
    END AS ProcessingDays,
    CASE
        WHEN o.shippedDate IS NOT NULL AND julianday(o.shippedDate) > julianday(o.requiredDate) THEN 1
        ELSE 0
    END AS IsLate
FROM orders o
JOIN orderdetails od ON o.orderNumber = od.orderNumber
JOIN products p ON od.productCode = p.productCode
WHERE o.status != 'Cancelled'
""",
conn)

Sales_KPI_5
```

6. Repeat Customer Rate : Measure customer loyalty by identifying customers who place multiple orders.

```python
Sales_KPI_6 = pd.read_sql("""
SELECT
    c.customerNumber,
    c.customerName,
    c.city,
    c.state,
    c.postalCode,
    c.country,
    c.salesRepEmployeeNumber,
    -- Overall Summary Metrics by Customer
    COUNT(DISTINCT o.orderNumber) AS total_orders_count,
    MIN(o.orderDate) AS first_order_date,
    MAX(o.orderDate) AS last_order_date,
    CASE
        WHEN COUNT(DISTINCT o.orderNumber) > 1 THEN 1
        ELSE 0
    END AS is_repeat_customer
FROM customers c
LEFT JOIN orders o ON c.customerNumber = o.customerNumber AND o.status != 'Cancelled'
GROUP BY
    c.customerNumber,
    c.customerName,
    c.city,
    c.state,
    c.postalCode,
    c.country,
    c.salesRepEmployeeNumber;
""",
conn)

Sales_KPI_6
```

7. Cancelled Sales

```python
Sales_KPI_7 = pd.read_sql("""
SELECT
    o.orderNumber,
    c.customerName as client,
    o.orderDate,
    od.quantityOrdered,
    od.priceEach,
    ROUND(od.quantityOrdered * od.priceEach, 2)  AS montant_commande,
    o.comments
FROM orders o
JOIN orderdetails od ON o.orderNumber   = od.orderNumber
JOIN customers c     ON c.customerNumber = o.customerNumber
WHERE o.status = 'Cancelled'
ORDER BY o.orderDate DESC;
""",
conn)

Sales_KPI_7
```

# Supply Chain

1. Stock Below Critical Threshold : Identify products with low stock levels to prevent stockouts.

```python
pd.read_sql("""
-- You can adjust the threshold (100 here) according to your needs.
SELECT
    productCode,
    productName,
    productLine,
    quantityInStock,
    buyPrice,
    MSRP,
    CASE
        WHEN quantityInStock < 25 THEN 'Urgent – Very Critical Stock'
        WHEN quantityInStock < 50 THEN 'Critical – Low Stock'
        WHEN quantityInStock < 100 THEN 'Warning – Moderate Stock'
        ELSE 'Stock suffisant'
    END AS Alert_Lv,
    ROUND(quantityInStock * buyPrice, 2) AS stock_values
FROM
    products
WHERE
    quantityInStock < 100
ORDER BY
    quantityInStock ASC;
""",
conn)
```

```python
SC_KPI_1 = pd.read_sql("""
-- You can adjust the threshold (100 here) according to your needs.
SELECT
    productCode,
    productName,
    productLine,
    quantityInStock,
    buyPrice,
    MSRP,
    CASE
        WHEN quantityInStock < 25 THEN 'Urgent – Very Critical Stock'
        WHEN quantityInStock < 50 THEN 'Critical – Low Stock'
        WHEN quantityInStock < 100 THEN 'Warning – Moderate Stock'
        ELSE 'Stock suffisant'
    END AS Alert_Lv,
    ROUND(quantityInStock * buyPrice, 2) AS stock_values
FROM
    products
ORDER BY
    quantityInStock ASC;
""",
conn)

SC_KPI_1
```

2. Average Order Processing Time + Orders Above Average Delivery Time

```python
pd.read_sql("""
SELECT
    ROUND(AVG(julianday(shippedDate) - julianday(orderDate)), 2) AS avg_processing_days
FROM orders
WHERE shippedDate IS NOT NULL;
""",
conn)
```

```python
SC_KPI_2 = pd.read_sql("""
WITH processing_times AS (
    SELECT
        orderNumber,
        customerNumber,
        orderDate,
        shippedDate,
        julianday(shippedDate) - julianday(orderDate) AS processing_days
    FROM orders
    WHERE shippedDate IS NOT NULL
)
SELECT
    orderNumber,
    customerNumber,
    orderDate,
    shippedDate,
    ROUND(processing_days, 1) AS processing_days
FROM processing_times
ORDER BY processing_days DESC;
""",
conn)

SC_KPI_2
```

3. Stock Sell-Through Rate : Measure how quickly inventory is sold for each product.

```python
SC_KPI_3 = pd.read_sql("""
SELECT
    p.productName,
    strftime('%Y', o.orderDate) AS annee,
    SUM(od.quantityOrdered) AS quantite_totalDesCommande,
    p.quantityInStock,
    ROUND(
        (SUM(od.quantityOrdered) * 1.0 / (SUM(od.quantityOrdered) + p.quantityInStock)) * 100,
        2
    ) AS taux_decoulement
FROM products p
JOIN orderdetails od ON p.productCode = od.productCode
JOIN orders o ON od.orderNumber = o.orderNumber
GROUP BY
    p.productName,
    p.quantityInStock,
    strftime('%Y', o.orderDate)
ORDER BY
    p.productName,
    annee DESC;
""",
conn)

SC_KPI_3
```

4. Late Delivery Rate : Identify logistics issues and improve delivery times.

```python
SC_KPI_4 = pd.read_sql("""
SELECT
    COUNT(*) AS total_orders,
    SUM(CASE WHEN julianday(shippedDate) > julianday(requiredDate) THEN 1 ELSE 0 END) AS late_orders,
    ROUND(
        100.0 * SUM(CASE WHEN julianday(shippedDate) > julianday(requiredDate) THEN 1 ELSE 0 END) / COUNT(*),
        2
    ) AS late_delivery_rate
FROM orders
WHERE shippedDate IS NOT NULL;
""",
conn)

SC_KPI_4
```

# Human Resources

1. Sales Representative Performance : Calculate the revenue generated by each sales representative.

```python
HR_KPI_1 = pd.read_sql("""
SELECT
    e.lastname || ' ' || e.firstname AS employees,
    strftime('%Y', p.paymentDate) AS annee,
    strftime('%m', p.paymentDate) AS mois,
    COALESCE(SUM(p.amount), 0) AS CA_montant
FROM employees e
LEFT JOIN customers c ON c.salesRepEmployeeNumber = e.employeeNumber
LEFT JOIN payments p ON p.customerNumber = c.customerNumber
WHERE e.jobTitle = 'Sales Rep'
GROUP BY
    e.lastname,
    e.firstname,
    strftime('%Y', p.paymentDate),
    strftime('%m', p.paymentDate)
ORDER BY
    annee DESC,
    mois DESC,
    CA_montant DESC,
    employees;
""",
conn)

HR_KPI_1
```

2. Orders-to-Payments Ratio by Sales Representative : Identify discrepancies between orders placed and payments received for each sales representative.

```python
HR_KPI_2 = pd.read_sql("""
WITH VentesParRep AS (
    -- Total Amount Due by Sales Representative
    SELECT
        c.salesRepEmployeeNumber,
        SUM(od.quantityOrdered * od.priceEach) AS total_du
    FROM orderdetails od
    JOIN orders o ON od.orderNumber = o.orderNumber
    JOIN customers c ON o.customerNumber = c.customerNumber
    GROUP BY c.salesRepEmployeeNumber
),
PaiementsParRep AS (
    -- Total Amount Paid by Sales Representative
    SELECT
        c.salesRepEmployeeNumber,
        SUM(p.amount) AS total_paye
    FROM payments p
    JOIN customers c ON p.customerNumber = c.customerNumber
    GROUP BY c.salesRepEmployeeNumber
)
SELECT
    e.lastName || ' ' || e.firstName AS employer,
    ROUND(COALESCE(v.total_du, 0), 2) AS total_du,
    ROUND(COALESCE(p.total_paye, 0), 2) AS total_paye,
    ROUND(
        CASE
            WHEN COALESCE(v.total_du, 0) > 0
            THEN (COALESCE(p.total_paye, 0) / v.total_du) * 100
            ELSE 0
        END, 2
    ) || ' %' AS ratio
FROM employees e
LEFT JOIN VentesParRep v ON e.employeeNumber = v.salesRepEmployeeNumber
LEFT JOIN PaiementsParRep p ON e.employeeNumber = p.salesRepEmployeeNumber
WHERE e.jobTitle = 'Sales Rep'
ORDER BY (COALESCE(p.total_paye, 0) / NULLIF(v.total_du, 0)) DESC;
""",
conn)

HR_KPI_2
```

3. Branch Performance : Measure the revenue generated by each branch.

```python
HR_KPI_3 = pd.read_sql("""
SELECT
    o.city,
    strftime('%Y', p.paymentDate) AS annee,
    strftime('%m', p.paymentDate) AS mois,
    ROUND(SUM(p.amount), 2) AS CA_montant
FROM payments p
JOIN customers c ON c.customerNumber = p.customerNumber
JOIN employees e ON c.salesRepEmployeeNumber = e.employeeNumber
JOIN offices o ON e.officeCode = o.officeCode
GROUP BY
    o.city,
    strftime('%Y', p.paymentDate),
    strftime('%m', p.paymentDate)
ORDER BY
    o.city,
    annee DESC,
    mois DESC,
    CA_montant DESC;
""",
conn)

HR_KPI_3
```
